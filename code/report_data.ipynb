## Check_data

We identified and corrected two unadjusted splits (EURO BANKS 80:1, PALLADIUM LVG 4:1) and two erroneous zero-volume prints (WORLD MIN VOL, ROBOTICS on 24/10/2025). 
The −64.5% move on COPPER LVG (31/07/2025) was verified as genuine and kept.

### Data quality: anomalies detected and corrections applied

1. EURO BANKS and PALLADIUM LVG: unadjusted share splits

Both ETFs went through a share split: each unit was divided into several smaller units, which changes the price per unit but not the value held. EURO BANKS had an 80:1 split on 26 October 2026 (from €349.80 to €4.43), and PALLADIUM LVG a 4:1 split on 19 December 2022 (from €153.98 to €35.50).

In principle, Yahoo Finance’s adjusted close (adj_close) accounts for splits. In practice, it only does so when Yahoo has recorded the split, which it often misses for European-listed ETFs. In our dataset, close and adj_close are identical on these dates, which confirms that no adjustment was made. Left uncorrected, these splits show up as fake losses of −98.7% and −77% in a single day, distorting returns, volatility, moving averages and drawdowns.

Correction: we divided close and adj_close before each split date by the split ratio (80 and 4). After correction, the moves become realistic: EURO BANKS gains about +1.3% on the split day instead of losing 98.7%.

2. WORLD MIN VOL and ROBOTICS: erroneous price on 24 October 2025

On 24 October 2025, both ETFs show an apparent jump of +16% and +18%, followed by an equivalent drop on the next trading day. Two signs show this is not a genuine market move: trading volume was zero that day (no trade took place at that price), and the price returned to its previous level the next day.

The gap is a factor of about 1.16–1.18, which matches the EUR/USD exchange rate at that time. The most likely explanation is that Yahoo recorded a USD price instead of the EUR price that day, as both ETFs are denominated in USD. This is a hypothesis, but it is consistent with the figures.

Correction: we replaced the close and adj_close values for that day with those of the previous trading day.

3. COPPER LVG: genuine extreme move on 31 July 2025 (kept)

Our automated check flags any daily move above 20% for manual review. It flagged a −64.5% drop on COPPER LVG on 31 July 2025. Unlike the cases above, this move is genuine. The US copper market experienced its steepest one-day price drop on record as COMEX copper futures plunged 22% following the White House announcement July 30 that refined copper would be excluded from the 50% import tariff on copper products. As COPPER LVG is a 3x leveraged product, a fall of about 22% in copper translates into a loss of roughly 65% for the ETF. Trading volume was also high that day, which is consistent with a real market event.

Decision: no correction; the data point was kept. This case illustrates why the 20% threshold is used as an alert rather than an automatic correction, and why leveraged ETFs carry a much higher risk than standard ETFs. A single day can erase most of their value.
