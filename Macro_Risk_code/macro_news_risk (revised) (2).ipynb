````python
import os, json
from pathlib import Path
import pandas as pd
import feedparser
from urllib.parse import quote

# Find the project folder automatically (the one with data/prices.csv), so it works on everyone's PC
here = Path.cwd()
for folder in [here, *here.parents]:
    if (folder / "data" / "prices.csv").exists():
        os.chdir(folder)
        break
print("Project folder:", Path.cwd())
PRICES_FILE    = Path("data/prices.csv")
PORTFOLIO_FILE = Path("trading/portfolio.csv")   # created later by the team; until then: all cash

# ---- Competition rules (FAQ) - do not change ----
MAX_ORDERS_PER_DAY = 5        # including sells
MIN_CASH_EUR       = 0.0      # no borrowing

# ---- Our risk rules: BALANCED investor profile (team decision) ----
# Logic: no single bet should cost the portfolio more than ~8-10% in a worst case
# (weight x largest 5-year drawdown, e.g. Semis 20% x -39% = -7.8%; 3x ETFs 10% x -80% = -8%).
MAX_WEIGHT_PER_ETF   = 0.20   # <<< ADAPT  1) max 20% of the portfolio in one ETF ...
EXEMPT_FROM_LIMITS   = ["OVERNIGHT CASH"]   #  ... except money market, which behaves like cash
MAX_LEVERAGED_WEIGHT = 0.10   # <<< ADAPT  2) max 10% in all leveraged/inverse ETFs together
MIN_DEFENSIVE_WEIGHT = 0.20   # <<< ADAPT  3) at least 20% in defensive assets (incl. uninvested cash)
DEFENSIVE_ETFS       = ["OVERNIGHT CASH", "EUR GOVT BOND", "GLOBAL AGG BOND"]  # <<< ADAPT
#   not defensive: HIGHY BOND (credit risk), US TREASURY 20Y (8.7% volatility, -36.5% max drawdown)
#   uninvested cash earns no interest in the competition -> keep the defensive part mainly in OVERNIGHT CASH
MAX_THEME_WEIGHT     = 0.40   # <<< ADAPT  4) max 40% in one theme (e.g. tech)
THEME_GROUPS = {"Frontier tech": "Equity / Tech"}   # <<< ADAPT  themes that move together count as one
UNDERLYING_THEME = {   # <<< ADAPT  leveraged ETFs count towards the theme they really track
    "NASDAQ 3X":     "Equity / Tech",
    "EURO STOXX 2X": "European equity",
    "COPPER LVG":    "Industrial & strategic metals",
    "PALLADIUM LVG": "Precious metals",
}   # inverse ETFs (S&P 500 SHORT, STOXX SHORT) reduce exposure, so they stay in "Leveraged & inverse"
MIN_ORDER_EUR        = 500.0  # <<< ADAPT  5) ignore tiny orders (we only have 5 per day)
#                                          6) no selling more than we hold (always on)

# Planning buffers for the Portfolio Manager (Francesco's safety margin): the AI plans with some room,
# so normal price moves before execution don't break the hard limits above, which stay the final check.
PLAN_MIN_DEFENSIVE = 0.25     # plan with at least 25% defensive (hard limit: 20%)
PLAN_MAX_THEME     = 0.35     # plan with at most 35% per theme (hard limit: 40%)
STARTING_CAPITAL     = 10_000.0    # official starting capital of the competition (EUR)
````

## 3. Risk rules – `check_risk()`

**Order format** (to agree with Francesco):
```python
[{"code": "NASDAQ", "action": "buy", "amount_eur": 1500},
 {"code": "GOLD",   "action": "sell", "amount_eur": 500}]
```
**Risk profile: Balanced**

*Competition rules (mandatory)*
1. Max **5 orders** per trading day, including sells – extra ones rejected
2. **No borrowing**: cash can never go below zero
3. If cash runs out, the **last buys are cut first**
4. **Sells are executed before buys**, so they can fund them
5. Orders are executed at the **previous day's closing price (T-1)**
6. Only ETFs from our **list of 50**

*Our risk rules (balanced profile)* – a buy that breaks a rule is **reduced to fit**, or rejected if less than the minimum order is left
1. Max **20%** of the portfolio in a single ETF (except **OVERNIGHT CASH**, which behaves like cash)
2. Max **10%** in leveraged and inverse ETFs combined
3. At least **20% in defensive assets**: OVERNIGHT CASH, EUR GOVT BOND, GLOBAL AGG BOND and uninvested cash
4. Max **40% in a single theme** (Equity / Tech and Frontier tech count as one; leveraged ETFs count towards the theme they track, e.g. NASDAQ 3X → tech)
5. Min order size of **€500**
6. **No selling more than we hold**

The Portfolio Manager plans with a **safety margin** (`PLAN_MIN_DEFENSIVE` 25%, `PLAN_MAX_THEME` 35%); these hard limits stay the final check.

Limits are measured on the portfolio value **before** the day's trades. After the trades, the function also lists **warnings** if a limit is broken by price moves (e.g. an ETF that grew above 20%), so we can decide whether to rebalance.

````python
THEME_OF = universe.set_index("code")["theme"].to_dict()

def etf_theme(code):
    """Theme used for the 40% cap and the macro check: the underlying exposure, with related themes merged."""
    t = UNDERLYING_THEME.get(code, THEME_OF.get(code))
    return THEME_GROUPS.get(t, t)

def check_risk_rules(orders, portfolio=None):
    portfolio = load_portfolio() if portfolio is None else portfolio
    px = last_close
    lev = universe.set_index("code")["leverage"].to_dict()
    themes = universe.set_index("code")["theme"].to_dict()
    holdings = portfolio[portfolio["code"] != "CASH"].set_index("code")["units"].to_dict()
    cash = float(portfolio.loc[portfolio["code"] == "CASH", "units"].sum())
    value = lambda c: holdings.get(c, 0) * px.get(c, 0)
    total = cash + sum(value(c) for c in holdings)          # portfolio value before trading
    approved, rejected = [], []

    # Competition rule 1: max 5 orders
    for o in orders[MAX_ORDERS_PER_DAY:]:
        rejected.append({**o, "reason": f"more than {MAX_ORDERS_PER_DAY} orders"})
    orders = orders[:MAX_ORDERS_PER_DAY]

    # Competition rule 6 + risk rule 5: ETF in our list, valid action, minimum order
    valid = []
    for o in orders:
        c, a, amt = o.get("code"), str(o.get("action", "")).lower(), float(o.get("amount_eur", 0))
        if c not in lev:
            rejected.append({**o, "reason": "ETF not in our list"})
        elif a not in ("buy", "sell"):
            rejected.append({**o, "reason": "action must be buy or sell"})
        elif amt < MIN_ORDER_EUR:
            rejected.append({**o, "reason": f"below {MIN_ORDER_EUR:.0f} EUR"})
        else:
            valid.append({"code": c, "action": a, "amount_eur": amt})

    # Competition rule 4 + risk rule 6: sells first, never more than we hold
    for o in [o for o in valid if o["action"] == "sell"]:
        held = value(o["code"])
        if held <= 0:
            rejected.append({**o, "reason": "we don't hold it"}); continue
        amt = min(o["amount_eur"], held)
        holdings[o["code"]] -= amt / px[o["code"]]
        cash += amt
        approved.append({**o, "amount_eur": round(amt, 2), "units": round(amt / px[o["code"]], 4),
                         "price_T-1": round(px[o["code"]], 4),
                         "note": "cut to what we hold" if amt < o["amount_eur"] else ""})

    # Helpers for the portfolio-level rules
    is_lev  = lambda c: lev.get(c, 1) != 1
    is_def  = lambda c: c in DEFENSIVE_ETFS
    theme   = etf_theme
    lev_value   = lambda: sum(value(c) for c in holdings if is_lev(c))
    def_value   = lambda: cash + sum(value(c) for c in holdings if is_def(c))   # uninvested cash counts
    theme_value = lambda t: sum(value(c) for c in holdings if theme(c) == t and c not in EXEMPT_FROM_LIMITS)

    # Buys in the proposed order (so a cash shortfall cuts the LAST buys - competition rule 3)
    for o in [o for o in valid if o["action"] == "buy"]:
        c, amt, why = o["code"], o["amount_eur"], []
        def cap(room, reason):
            nonlocal amt
            if amt > room:
                amt = max(room, 0)
                why.append(reason)
        if c not in EXEMPT_FROM_LIMITS:
            cap(MAX_WEIGHT_PER_ETF * total - value(c), f"max {MAX_WEIGHT_PER_ETF:.0%} per ETF")          # risk rule 1
            cap(MAX_THEME_WEIGHT * total - theme_value(theme(c)),
                f"max {MAX_THEME_WEIGHT:.0%} in theme '{theme(c)}'")                                     # risk rule 4
        if is_lev(c):
            cap(MAX_LEVERAGED_WEIGHT * total - lev_value(), f"max {MAX_LEVERAGED_WEIGHT:.0%} in leveraged ETFs")  # risk rule 2
        if not is_def(c):   # paying cash for a non-defensive ETF lowers the defensive share
            cap(def_value() - MIN_DEFENSIVE_WEIGHT * total, f"min {MIN_DEFENSIVE_WEIGHT:.0%} defensive")  # risk rule 3
        cap(cash - MIN_CASH_EUR, "not enough cash")                                                      # competition rule 2
        if amt < MIN_ORDER_EUR:                                                                           # risk rule 5
            rejected.append({**o, "reason": "; ".join(why) or "too small"}); continue
        holdings[c] = holdings.get(c, 0) + amt / px[c]
        cash -= amt
        approved.append({**o, "amount_eur": round(amt, 2), "units": round(amt / px[c], 4),
                         "price_T-1": round(px[c], 4), "note": "; ".join(why)})

    # Portfolio after the trades + warnings (limits can also be broken later by price moves)
    weights = {c: float(round(value(c) / total, 3)) for c in holdings if holdings[c] > 1e-9}
    weights["CASH"] = float(round(cash / total, 3))
    summary = {"defensive": float(round(def_value() / total, 3)),
               "uninvested_cash": float(round(cash / total, 3)),
               "leveraged": float(round(lev_value() / total, 3)),
               "themes": {t: float(round(theme_value(t) / total, 3))
                          for t in sorted({theme(c) for c in holdings if holdings[c] > 1e-9 and c not in EXEMPT_FROM_LIMITS})}}
    warnings = [f"{c} is {w:.0%} of the portfolio (max {MAX_WEIGHT_PER_ETF:.0%})"
                for c, w in weights.items() if c != "CASH" and c not in EXEMPT_FROM_LIMITS and w > MAX_WEIGHT_PER_ETF + 0.005]
    warnings += [f"theme '{t}' is {w:.0%} (max {MAX_THEME_WEIGHT:.0%})"
                 for t, w in summary["themes"].items() if w > MAX_THEME_WEIGHT + 0.005]
    if summary["leveraged"] > MAX_LEVERAGED_WEIGHT + 0.005:
        warnings.append(f"leveraged ETFs are {summary['leveraged']:.0%} (max {MAX_LEVERAGED_WEIGHT:.0%})")
    if summary["defensive"] < MIN_DEFENSIVE_WEIGHT - 0.005:
        warnings.append(f"defensive assets are {summary['defensive']:.0%} (min {MIN_DEFENSIVE_WEIGHT:.0%})")
    if summary["uninvested_cash"] > 0.10:
        warnings.append(f"{summary['uninvested_cash']:.0%} is uninvested cash, which earns no interest: "
                        "consider OVERNIGHT CASH")
    return {"approved": approved, "rejected": rejected, "cash_after_eur": round(cash, 2),
            "weights_after": weights, "summary": summary, "warnings": warnings}
````

## 4. ✏️ Macro/News – headlines per topic

Free source: **Google News RSS** (last 7 days). Each headline gets an **id** (e.g. `IR2`, `GEO5`) so the AI can cite it and we can check every claim.
Headlines are sorted by date and title, so the same news always reaches the AI in the same order (more consistent answers).

````python
import time

NEWS_TOPICS = {   # topic: (headline id prefix, search words)                  # <<< ADAPT
    "interest_rates": ("IR",  "ECB OR Federal Reserve interest rates"),
    "ecb_europe":     ("EU",  "ECB Lagarde euro zone"),
    "geopolitics":    ("GEO", "geopolitical risk markets"),
    "commodities":    ("COM", "oil OR copper OR gold prices"),
}
N_HEADLINES = 8                                                              # <<< ADAPT

def get_headlines(query, n=N_HEADLINES):
    """Latest headlines (last 7 days) as (date, title), newest first and in a fixed order."""
    url = f"https://news.google.com/rss/search?q={quote(query)}+when:7d&hl=en&gl=US&ceid=US:en"
    try:
        entries = feedparser.parse(url).entries[:n]
    except Exception as e:
        print(f"news unavailable ({type(e).__name__})")
        return []
    items = []
    for e in entries:
        t = e.get("published_parsed")
        items.append((time.strftime("%Y-%m-%d", t) if t else "", e.get("title", "")))
    return sorted(items, reverse=True)

def get_all_headlines():
    """All topics together: {headline id: "title (date)"}, e.g. {"IR1": "Fed signals ... (2026-11-10)"}."""
    out = {}
    for topic, (prefix, query) in NEWS_TOPICS.items():
        for i, (d, title) in enumerate(get_headlines(query), 1):
            out[f"{prefix}{i}"] = f"{title} ({d})"
    return out

headlines = get_all_headlines()
for hid, text in headlines.items():
    print(f"[{hid}] {text}")
if not headlines:
    print("No headlines found - check the internet connection.")
````

## 5b. Macro/News – AI view per theme (structured)

The **code** collects and numbers the headlines; the **AI** turns them into facts and one view per theme. The prompt was improved after Francesco's review of the first version:

| Problem found | Fix in the prompt / code |
|---|---|
| Money market linked to the **Fed**, but OVERNIGHT CASH follows the **ECB** | A fixed *market context*: all ETFs trade in EUR; money market and EUR bonds depend on the ECB |
| Generalisations ("some officials" for one official) | Rule: cite the headline id next to every fact; never generalise |
| Weak inferences (e.g. ECB leadership → energy transition) | Rule: link a headline to a theme only through a direct mechanism, otherwise neutral |
| Same theme positive in one topic and negative in another | **One net view per theme**, across all topics |
| Views flipping between runs with the same headlines | Fixed headline order, strict rules, "mixed or weak evidence → neutral", JSON output |
| Allocation advice mixed in | Rule: no buy/sell advice – that is the Portfolio Manager's job |
| "Fixed income / Bonds" mixed US and EUR bonds, so a Fed headline could turn EUR GOVT BOND negative | For the macro view, bonds are split into **EUR bonds** (EUR GOVT, GLOBAL AGG, HIGHY) and **US bonds** (US TREASURY 20Y). The risk limits keep the original themes |

**Output** of `get_macro_news()`:
- `theme_views`: `{theme: {"view": "positive/neutral/negative", "reason": ..., "sources": ["IR2", ...]}}` – every theme gets exactly one view (neutral if not mentioned). **The Portfolio Manager reads this.** The old key "Fixed income / Bonds" is kept for compatibility and copies the EUR bonds view.
- `etf_macro_views(code, theme_views)`: the views that matter for **one ETF** – e.g. QUANTUM and SPACE look at Frontier tech and tech, BEL 20 at Home market and European equity, and the two inverse ETFs take the **opposite** of the market they short.
- `facts`, `overall`, `headlines` (the raw numbered headlines, to check every claim) and `summary` (readable text for humans).
- `macro_conflicts(orders, theme_views)` lists orders that go **against** the macro view (buying with a negative view or selling with a positive one), so Python can flag them automatically.

````python
import json, re

# Macro themes = the data themes, except bonds, which are split by the interest rates that drive them.
# (The risk limits in section 3 keep using the data themes.)
MACRO_THEME_SPLIT = {                                                        # <<< ADAPT
    "EUR GOVT BOND":   "EUR bonds",
    "GLOBAL AGG BOND": "EUR bonds",     # hedged to EUR
    "HIGHY BOND":      "EUR bonds",     # EUR high-yield corporate bonds
    "US TREASURY 20Y": "US bonds",
}

def macro_theme(code):
    """Theme used by the macro agent for one ETF."""
    return MACRO_THEME_SPLIT.get(code, THEME_OF.get(code))

THEMES = sorted({macro_theme(c) for c in universe["code"]})
THEME_ETFS = {t: sorted(c for c in universe["code"] if macro_theme(c) == t) for t in THEMES}
VIEWS = ("positive", "neutral", "negative")

MARKET_CONTEXT = """- All our ETFs trade in EUR.
- OVERNIGHT CASH (money market) follows the ECB short-term rate (EUR STR), not the Fed.
- "EUR bonds" (EUR GOVT BOND, GLOBAL AGG BOND hedged to EUR, HIGHY BOND) depend on ECB / euro-area
  interest rates and credit spreads, not on the Fed.
- "US bonds" (US TREASURY 20Y) depend on the Fed and US long-term rates.
- The Fed also matters for US equity, the US dollar and global risk appetite.
- Leveraged ETFs follow their underlying index (e.g. NASDAQ 3X = US tech, 3 times the daily move)."""  # <<< ADAPT

def get_llm():
    """Connects to Gemini. Returns None if the settings are missing."""
    if not all(os.environ.get(k) for k in ("LLM_BASE_URL", "LLM_MODEL", "LLM_API_KEY")):
        return None
    from langchain_openai import ChatOpenAI
    return ChatOpenAI(base_url=os.environ["LLM_BASE_URL"], api_key=os.environ["LLM_API_KEY"],
                      model=os.environ["LLM_MODEL"], temperature=0, timeout=90, max_retries=1)

def _parse_json(text):
    """Takes the JSON object out of the AI answer (also if it is wrapped in ```json ... ```)."""
    m = re.search(r"\{.*\}", text or "", re.S)
    try:
        return json.loads(m.group(0)) if m else None
    except json.JSONDecodeError:
        return None

def _clean_views(raw):
    """Exactly one valid view per theme: unknown themes or views are dropped, missing themes are neutral."""
    views = {t: {"view": "neutral", "reason": "no direct evidence in the headlines", "sources": []} for t in THEMES}
    for v in (raw or {}).get("theme_views", []):
        t, view = v.get("theme"), str(v.get("view", "")).lower()
        if t in views and view in VIEWS:
            views[t] = {"view": view, "reason": v.get("reason", ""), "sources": list(v.get("sources", []))}
    return views

def _with_compat(views):
    """Keeps the old key "Fixed income / Bonds" for code written before the split.
    It copies the EUR bonds view (3 of our 4 bond ETFs are in EUR); new code should use etf_macro_views()."""
    v = dict(views)
    eur = views.get("EUR bonds", {"view": "neutral", "reason": "", "sources": []})
    v["Fixed income / Bonds"] = {**eur, "reason": f"same as EUR bonds: {eur.get('reason', '')}"}
    return v

def format_summary(data, views):
    """Readable text for humans, built from the structured answer."""
    lines = ["FACTS (from the headlines)"]
    lines += [f"- {f.get('text', '')} [{', '.join(f.get('sources', []))}]" for f in data.get("facts", [])]
    lines.append("\nNET VIEW PER THEME (inference)")
    for t, v in views.items():
        if v["sources"]:
            lines.append(f"- {t}: {v['view'].upper()} - {v['reason']} [{', '.join(v['sources'])}]")
    neutral = [t for t, v in views.items() if not v["sources"]]
    if neutral:
        lines.append(f"- No direct evidence (neutral): {', '.join(neutral)}")
    lines.append(f"\nOVERALL\n{data.get('overall', '')}")
    return "\n".join(lines)

def get_macro_news(retries=2, pause=20):
    """Collects the headlines and asks the AI for facts + one net view per theme (one request)."""
    headlines = get_all_headlines()
    empty = {"facts": [], "overall": "", "theme_views": _with_compat(_clean_views(None)), "headlines": headlines}
    if not headlines:
        return {**empty, "summary": "[no headlines]"}
    llm = get_llm()
    if llm is None:
        return {**empty, "summary": "[AI not configured]"}

    numbered = "\n".join(f"[{h}] {t}" for h, t in headlines.items())
    prompt = f"""You are the macro analyst of a team managing a EUR ETF portfolio for the next 2 weeks.

Headlines from the last 7 days, each with an id:
{numbered}

Market context (always true, use it):
{MARKET_CONTEXT}

Our ETF themes (use these exact names) and the ETFs in each: {json.dumps(THEME_ETFS)}

Rules:
1. Use ONLY the headlines above. Do not invent numbers, dates, names or facts.
2. Cite the headline id(s) next to every fact and every view, e.g. ["IR2"].
3. Do not generalise: if a headline mentions one official or one country, say so.
4. Keep facts (what the headlines say) separate from views (what it means for a theme).
5. Link a headline to a theme only through a direct, well-known mechanism
   (e.g. higher ECB rates -> lower EUR bond prices). If the link is indirect or uncertain, leave the theme out.
6. Give ONE net view per theme, weighing all topics together. If the evidence is mixed or weak, use "neutral".
7. No allocation or trading advice (no buy, sell, overweight, underweight): that is the Portfolio Manager's job.

Answer ONLY with JSON in exactly this format:
{{"facts": [{{"text": "...", "sources": ["IR1"]}}],
 "theme_views": [{{"theme": "<exact theme name>", "view": "positive|neutral|negative", "reason": "<one line>", "sources": ["IR1"]}}],
 "overall": "<2-3 sentences on the macro picture, no allocation advice>"}}"""                     # <<< ADAPT prompt

    for attempt in range(retries + 1):
        try:
            data = _parse_json(llm.invoke(prompt).content)
            if data:
                views = _clean_views(data)
                summary = format_summary(data, views)
                return {"facts": data.get("facts", []), "overall": data.get("overall", ""),
                        "theme_views": _with_compat(views), "headlines": headlines, "summary": summary}
            print("AI answer was not valid JSON, trying again...", flush=True)
        except Exception as e:
            print(f"AI busy ({type(e).__name__}), waiting {pause}s and trying again...", flush=True)
            time.sleep(pause)
    return {**empty, "summary": "[AI unavailable - see raw headlines]"}

# Which macro views matter for an ETF (+1 = same direction, -1 = opposite, for inverse ETFs)
MACRO_LOOKUP = {                                                              # <<< ADAPT
    "QUANTUM":       [("Frontier tech", +1), ("Equity / Tech", +1)],
    "SPACE":         [("Frontier tech", +1), ("Equity / Tech", +1)],
    "BEL 20":        [("Home market", +1), ("European equity", +1)],
    "NASDAQ 3X":     [("Equity / Tech", +1)],
    "EURO STOXX 2X": [("European equity", +1)],
    "COPPER LVG":    [("Industrial & strategic metals", +1)],
    "PALLADIUM LVG": [("Precious metals", +1)],
    "S&P 500 SHORT": [("US equity", -1)],         # inverse: gains when US equity falls
    "STOXX SHORT":   [("European equity", -1)],   # inverse: gains when European equity falls
}   # every other ETF: its own macro theme, same direction
FLIP = {"positive": "negative", "negative": "positive", "neutral": "neutral"}

def etf_macro_views(code, theme_views):
    """The macro views that matter for one ETF, already turned the right way for inverse ETFs."""
    out = []
    for theme, sign in MACRO_LOOKUP.get(code, [(macro_theme(code), +1)]):
        v = theme_views.get(theme)
        if v:
            out.append({"theme": theme, "view": v["view"] if sign > 0 else FLIP[v["view"]],
                        "reason": v.get("reason", ""), "inverse": sign < 0})
    return out

def macro_conflicts(orders, theme_views):
    """Orders that go against the macro view. Prudent: one opposite view is enough to flag an order."""
    out = []
    for o in orders:
        for v in etf_macro_views(o["code"], theme_views):
            if (o["action"] == "buy" and v["view"] == "negative") or (o["action"] == "sell" and v["view"] == "positive"):
                note = " (inverse ETF, so the opposite of the theme view)" if v["inverse"] else ""
                out.append(f"{o['action'].upper()} {o['code']}: macro view on '{v['theme']}' gives "
                           f"{v['view']}{note} ({v['reason']})")
    return out

macro_view = get_macro_news()
print(macro_view["summary"])
````
